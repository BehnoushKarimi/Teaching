# Classes and OOP — Challenging Practice Activities


## Activity 1 — Product Profit Analyzer 

Create a class called `Product`.

### Attributes
- `name`
- `selling_price`
- `cost_price`
- `units_sold`

### Methods
- `__init__(...)`
- `revenue()` → returns total revenue
- `total_cost()` → returns total cost
- `profit()` → returns total profit
- `performance()` → returns:
  - `"Excellent"` if profit is at least 5000
  - `"Good"` if profit is at least 1000
  - `"Low"` if profit is positive but below 1000
  - `"Loss"` if profit is negative
  - `"Break-even"` if profit is exactly 0
- `__str__()` → returns a readable summary

Create at least three product objects and compare their results.


## Activity 2 — Employee Performance and Bonus 

Create an `Employee` class.

### Attributes
- `name`
- `monthly_salary`
- `sales`
- `target`

### Methods
- `target_percentage()` → returns the percentage of the target achieved
- `bonus()`:
  - target achievement below 80% → no bonus
  - 80% to below 100% → 5% of monthly salary
  - 100% to below 120% → 10%
  - 120% or more → 20%
- `total_payment()` → salary + bonus
- `__str__()` → readable summary

If `target` is `0`, `target_percentage()` should return `0`.

Create several employees and identify which employee receives the highest total payment.


## Activity 3 — Shopping Cart with a Dictionary 

Create a `ShoppingCart` class.

### Attributes
- `customer_name`
- `items` → an empty dictionary when the object is created

The dictionary should store:

```python
product_name: total_price_for_that_product
```

### Methods
- `add_item(product_name, price, quantity)`
  - calculate `price * quantity`
  - add it to the product's current total if the product already exists
- `remove_item(product_name)`
  - remove the product if it exists
- `total()` → total value of the cart
- `largest_item()` → name of the product with the largest total value
- `__str__()` → customer name, number of different products, and cart total

Do not use `sum()` or `max()` for `total()` and `largest_item()`.


## Activity 4 — Inventory Across Several Stores 

Create a class called `StoreInventory`.

Each store object should have:

### Instance attributes
- `store_name`
- `stock` → an empty dictionary

The dictionary stores:

```python
product_name: quantity
```

### Class variable
- `total_units` → total number of units across **all** store objects

### Methods
- `add_stock(product, quantity)`
- `sell(product, quantity)`
  - sell only if enough stock exists
- `store_units()` → total units in this particular store
- `__str__()`

### Class method
- `total_company_stock()` → returns `total_units`

Create two stores, add and sell products, and show both store-level and company-level stock.

## Activity 6 — Business Utility Static Methods 

Create a class called `BusinessMath`.

It does **not** need `__init__()` because it stores no object data.

Create these static methods:

- `profit(revenue, cost)`
- `profit_margin(revenue, cost)`
- `growth_rate(old_value, new_value)`
- `discounted_price(price, discount_percent)`

Rules:
- if revenue is `0`, `profit_margin()` returns `0`
- if old value is `0`, `growth_rate()` returns `0`
- discount percent should only work from `0` to `100`; otherwise return `"Invalid discount"`

Call the methods directly using the class name.


## Activity 7 — Customer Credit Limit with Encapsulation

## Activity 9 — Loan Application Decision 

Create a class called `LoanApplication`.

### Attributes
- `applicant`
- `monthly_income`
- `monthly_debt`
- `requested_amount`
- `credit_score`

### Methods
- `debt_ratio()` → monthly debt / monthly income × 100
- `decision()`:
  - `"Rejected"` if income is 0
  - `"Rejected"` if credit score is below 550
  - `"Rejected"` if debt ratio is above 50%
  - `"Review"` if requested amount is more than 12 times monthly income
  - otherwise `"Approved"`
- `__str__()` → readable result

Create at least four applications that produce different decisions.

## Activity 12 — Mini Subscription Business System 

Create a `Subscription` class.

### Class variable
- `tax_rate = 0.19`

### Instance attributes
- `customer`
- `monthly_price`
- `months`
- `active`

### Methods
- `subtotal()` → monthly price × months
- `tax()` → subtotal × class tax rate
- `total()` → subtotal + tax
- `cancel()` → changes `active` to `False`
- `status()` → `"Active"` or `"Cancelled"`
- `__str__()`

### Class method
- `change_tax_rate(new_rate)`
  - only allow a value from `0` to `1`

### Static method
- `valid_months(months)` → returns `True` only when months is between 1 and 36

### Final challenge
Create several subscriptions and store them in a list.

Using a loop:
- calculate total expected revenue from **active** subscriptions
- count cancelled subscriptions
- find the active subscription with the highest total value

Then change the class tax rate and show that the objects use the new shared rate.
