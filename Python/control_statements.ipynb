# Control Statements — More Challenging Activities

These activities use the same Python knowledge you have already seen:
- expressions and basic data types
- `if`, `elif`, and `else`
- `for` and `while` loops
- strings, lists, and dictionaries
- comparison and logical operators

The problems are a little more challenging because you need to combine these ideas.

**Do not use functions, classes, list comprehensions, or other new Python topics.**


## 1. Number Analyzer

Write a Python program that takes an integer from the user and prints:

1. whether the number is **positive**, **negative**, or **zero**
2. whether it is **even** or **odd**
3. whether it is divisible by **3**
4. whether it is divisible by **both 2 and 3**

For example, if the user enters `12`, the program could print:

```text
12 is positive
12 is even
12 is divisible by 3
12 is divisible by both 2 and 3
```

Think carefully about the case when the number is `0`.


## 2. Valid Grade and Letter Grade

Write a Python program that asks the user for a grade from `0` to `100`.

First, check whether the grade is valid.

- If the grade is less than `0` or greater than `100`, print `"Invalid grade"`.
- Otherwise, print the letter grade:

```text
90–100  -> A
80–89   -> B
70–79   -> C
60–69   -> D
0–59    -> F
```

For example:

```text
Enter grade: 84
B
```

Be careful about the order of your `if` / `elif` conditions.


## 3. Ticket Price

A cinema uses the following ticket prices:

- younger than 5: free
- 5 to 12 years old: 5 euros
- 13 to 17 years old: 8 euros
- 18 to 64 years old: 12 euros
- 65 or older: 7 euros

Write a program that takes the user's age and prints the ticket price.

If the entered age is negative, print `"Invalid age"`.

### Extra condition

If the person is a student and is between 18 and 25 years old, the ticket costs `9 euros` instead of `12 euros`.

You may ask the user to enter `"yes"` or `"no"` for student status.


## 4. Guess the Number — Three Attempts

Create a predefined secret number, for example:

```python
secret_number = 17
```

Give the user **three attempts** to guess it.

After each wrong guess:

- print `"Too small"` if the guess is smaller than the secret number
- print `"Too large"` if the guess is larger than the secret number

If the user guesses correctly, print `"Correct!"` and stop asking for more guesses.

If all three attempts are used without a correct answer, print:

```text
You did not guess the number.
```

Try to solve this using a loop and control statements.


## 5. Count Different Types of Values

Given a list such as:

```python
values = [4, "hello", 2.5, True, "Python", 8, False, 3.7]
```

Write a program that counts how many elements are:

- strings
- integers
- floats
- booleans

At the end, print all four counts.

Example idea:

```text
Strings: 2
Integers: ...
Floats: ...
Booleans: ...
```

**Important:** In Python, `True` and `False` can behave like integers in some situations. Think about the order in which you check the data types.


## 6. Sum and Count Numbers Divisible by 3 or 5

Given a list of integers, for example:

```python
numbers = [4, 6, 10, 11, 15, 18, 20, 22, 30]
```

Find all numbers that are divisible by **3 or 5**.

Your program should calculate and print:

1. how many such numbers there are
2. their total sum

For the example above, the selected numbers are:

```text
6, 10, 15, 18, 20, 30
```

Be careful: a number such as `15` is divisible by both 3 and 5, but it should only be counted once.


## 7. Find the Largest and Smallest Numbers Without `max()` or `min()`

Given a non-empty list such as:

```python
numbers = [14, 7, 31, 9, 22, 3]
```

Write a program that finds both:

- the largest number
- the smallest number

Do **not** use `max()` or `min()`.

Use variables, a loop, and control statements.

For the example above:

```text
Largest: 31
Smallest: 3
```


## 8. Longest Word — Including Ties

Ask the user to enter a sentence.

Find the length of the longest word, and print **all words** that have that length.

For example:

```text
Input: Python feels really simple today
```

Both `"Python"` and `"really"` have 6 letters, so the program should print both of them.

You can first split the sentence into words and then use loops and control statements.


## 9. Divisors and Number Classification

Ask the user for a positive integer.

First, find all of its divisors and store them in a list.

Then classify the number:

- If it has exactly two divisors, print `"Prime"`.
- Otherwise, print `"Not prime"`.

For example:

```text
Number: 7
Divisors: [1, 7]
Prime
```

```text
Number: 12
Divisors: [1, 2, 3, 4, 6, 12]
Not prime
```

If the user enters `0` or a negative number, print `"Please enter a positive integer"`.


## 10. Most Frequent Element

Given a list such as:

```python
items = ["red", "blue", "red", "green", "blue", "red", "green"]
```

Create a dictionary that stores the frequency of every element.

Then find which element appears the most.

For the example above:

```text
{'red': 3, 'blue': 2, 'green': 2}
Most frequent: red
```

Do not assume in advance which element will be the most frequent.


## 11. Simple ATM Decision

Suppose:

```python
balance = 500
```

Ask the user how much money they want to withdraw.

Apply these rules:

- the amount must be greater than `0`
- the amount must not be greater than the balance
- the amount must be divisible by `10`

If all conditions are satisfied:

1. subtract the amount from the balance
2. print `"Withdrawal successful"`
3. print the new balance

Otherwise, print a useful message explaining why the withdrawal cannot happen.

Example:

```text
Balance: 500
Withdrawal: 120
Withdrawal successful
New balance: 380
```


## 12. Mini Challenge — Analyze a List of Scores

Given:

```python
scores = [76, 41, 89, 95, 58, 73, 100, 67, 32, 84]
```

Write one program that finds:

- how many students passed (`score >= 60`)
- how many students failed
- the highest score, without using `max()`
- the lowest score, without using `min()`
- the sum of all passing scores

Finally, print all five results.

### Extra challenge

Also count how many scores are:

- `90` or higher
- between `80` and `89`
- between `70` and `79`
- between `60` and `69`
- below `60`

This activity combines several control-statement ideas from the session.
