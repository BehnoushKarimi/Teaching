- `"divisible_by_3"`: `True` or `False`

Examples:

```python
number_profile(246)
# {'sign': 'positive', 'even': True, 'divisible_by_3': True, 'digit_count': 3}

number_profile(-15)
# {'sign': 'negative', 'even': False, 'divisible_by_3': True, 'digit_count': 2}
```
