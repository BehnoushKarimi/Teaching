## Activity 1 — Business Loan Calculator with `math` 


A company wants a simple loan-analysis tool.

```python
loan_summary(principal, annual_rate, years)
```

```text
future_value = principal * (1 + annual_rate) ** years
```

The function should return a dictionary containing:

- `future_value`
- `total_interest`
- `average_interest_per_year`
- `rounded_future_value` rounded **up** to the next whole euro using `math.ceil()`

- `principal` must be greater than 0
- `annual_rate` must be between 0 and 1
- `years` must be greater than 0
- otherwise return `"Invalid data"`

Test at least three cases.
